```python
"""
Life Expectancy Terminal App (Python)
Based on: Appendix 1 - Model Life Table, National, Males (2021 PHC)

DATA ERRORS FIXED:
  - Age groups "1-4", "5-9", "10-14" were stored as Excel date serial
    numbers (46113, 46151, 46309) — corrected to proper age labels.
  - Age group "65-65" was a typo — corrected to "65-69".
"""

import pandas as pd

# ── 1. Build the cleaned life expectancy table ─────────────────────────────
# Each row: (age_min, age_max, life_expectancy_in_years)
# age_max = 999 for the open-ended "80+" group

data = {
    "Age Group": ["0", "1-4", "5-9", "10-14", "15-19", "20-24",
                  "25-29", "30-34", "35-39", "40-44", "45-49",
                  "50-54", "55-59", "60-64", "65-69", "70-74",
                  "75-79", "80+"],
    "Age Min":   [0,  1,  5, 10, 15, 20, 25, 30, 35, 40, 45,
                  50, 55, 60, 65, 70, 75, 80],
    "Age Max":   [0,  4,  9, 14, 19, 24, 29, 34, 39, 44, 49,
                  54, 59, 64, 69, 74, 79, 999],   # 999 = open-ended (80+)
    "Life Expectancy (yrs)": [
        63.28, 65.26, 62.14, 57.61, 52.92, 48.50,
        44.29, 40.02, 35.72, 31.32, 27.13,
        23.07, 19.31, 15.71, 12.54,  9.88,
         7.92,  7.40
    ]
}

df = pd.DataFrame(data)


# ── 2. Lookup function ──────────────────────────────────────────────────────

def get_life_expectancy(age):
    """Return the life expectancy for a given age."""
    row = df[(df["Age Min"] <= age) & (df["Age Max"] >= age)]
    if row.empty:
        return None
    return row.iloc[0]["Life Expectancy (yrs)"], row.iloc[0]["Age Group"]


# ── 3. Main terminal loop ───────────────────────────────────────────────────

def main():
    print("=" * 50)
    print("  Life Expectancy Calculator (Males, Ghana 2021)")
    print("=" * 50)
    print("Type 'quit' at any time to exit.\n")

    while True:
        user_input = input("Enter your age: ").strip()

        # Allow the user to exit
        if user_input.lower() in ("quit", "exit", "q"):
            print("Goodbye!")
            break

        # Validate: must be a whole number
        if not user_input.isdigit():
            print("  ⚠  Please enter a valid whole number (e.g. 25).\n")
            continue

        age = int(user_input)

        # Reasonable age range check
        if age < 0 or age > 120:
            
            print("  ⚠  Age must be between 0 and 120.\n")
            continue

        result = get_life_expectancy(age)

        if result is None:
            print("  ⚠  No data found for that age.\n")
        else:
            life_exp, age_group = result
            print(f"\n  Age group : {age_group}")
            print(f"  ➜  You have a life expectancy of {life_exp} more years.\n")


if __name__ == "__main__":
    main()
```
